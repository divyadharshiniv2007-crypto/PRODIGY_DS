```python
# ============================================
# PRODIGY INFOTECH - DATA SCIENCE
# TASK 01: Population Data Visualization
# ============================================

print("\nObjective:")
print("""
To visualize the distribution of population data using a histogram and a bar chart, and identify the countries with the highest populations in 2025.
""")

print("\nDataset Description:")
print("""
The dataset contains population information for countries and regions from the World Bank's World Development Indicators. The analysis focuses on population data for the year 2025.
""")


# 1. Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import zipfile
import os

# --------------------------------------------
# 2. Upload the ZIP file
# --------------------------------------------

from google.colab import files

uploaded = files.upload()

# Get the uploaded ZIP filename
zip_filename = list(uploaded.keys())[0]

print("Uploaded file:", zip_filename)


# --------------------------------------------
# 3. Extract the ZIP file
# --------------------------------------------

extract_folder = "task1_data"

with zipfile.ZipFile(zip_filename, "r") as zip_ref:
    zip_ref.extractall(extract_folder)

print("ZIP file extracted successfully!")


# --------------------------------------------
# 4. Find the CSV file automatically
# --------------------------------------------

csv_file = None

for root, directories, filenames in os.walk(extract_folder):
    for filename in filenames:
        if filename.lower().endswith(".csv"):
            csv_file = os.path.join(root, filename)
            break
    if csv_file:
        break

if csv_file is None:
    raise FileNotFoundError("No CSV file was found inside the ZIP file.")

print("CSV file found:", csv_file)


# --------------------------------------------
# 5. Load the dataset
# --------------------------------------------

df = pd.read_csv(csv_file, skiprows=4)

print("\nDataset loaded successfully!")


# --------------------------------------------
# 6. Display first 5 rows
# --------------------------------------------

print("\nFirst 5 rows:")
display(df.head())


# --------------------------------------------
# 7. Display dataset information
# --------------------------------------------

print("\nDataset Information:")
df.info()


# --------------------------------------------
# 8. Display dataset shape
# --------------------------------------------

print("\nDataset Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# --------------------------------------------
# 9. Display column names
# --------------------------------------------

print("\nColumn Names:")
print(df.columns.tolist())


# --------------------------------------------
# 10. Remove unnecessary unnamed columns
# --------------------------------------------

unnamed_columns = [
    column for column in df.columns
    if str(column).startswith("Unnamed")
]

if unnamed_columns:
    df = df.drop(columns=unnamed_columns)

print("\nUnnecessary columns removed.")


# --------------------------------------------
# 11. Check missing values in 2025
# --------------------------------------------

print("\nMissing values in 2025:")
print(df["2025"].isnull().sum())


# --------------------------------------------
# 12. Create a clean population column
# --------------------------------------------

population_2025 = df["2025"].dropna()


# --------------------------------------------
# 13. HISTOGRAM
# Population Distribution in 2025
# --------------------------------------------

plt.figure(figsize=(10, 6))

plt.hist(population_2025, bins=20)

plt.xlabel("Population")
plt.ylabel("Number of Countries/Regions")
plt.title("Population Distribution in 2025")

plt.show()


# --------------------------------------------
# 14. HISTOGRAM WITH LOG SCALE
# Better visualization because population
# values vary greatly
# --------------------------------------------

plt.figure(figsize=(10, 6))

plt.hist(population_2025, bins=20)

plt.xscale("log")

plt.xlabel("Population (Log Scale)")
plt.ylabel("Number of Countries/Regions")
plt.title("Population Distribution in 2025")

plt.show()


# --------------------------------------------
# 15. TOP 10 MOST POPULATED COUNTRIES IN 2025
# --------------------------------------------

# Install pycountry to identify actual countries
!pip install pycountry -q

import pycountry

# Get ISO 3-letter country codes
country_codes = {
    country.alpha_3
    for country in pycountry.countries
}

# Keep only rows whose Country Code belongs to an actual country
countries = df[df["Country Code"].isin(country_codes)].copy()

# Remove rows where 2025 population is missing
countries = countries.dropna(subset=["2025"])

# Get the 10 most populated countries
top10 = countries.nlargest(10, "2025")

print("Top 10 Most Populated Countries in 2025:")
display(top10[["Country Name", "Country Code", "2025"]])


# --------------------------------------------
# 16. BAR CHART
# --------------------------------------------

plt.figure(figsize=(12, 6))

plt.bar(top10["Country Name"], top10["2025"])

plt.xlabel("Country")
plt.ylabel("Population")
plt.title("Top 10 Most Populated Countries in 2025")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()

# --------------------------------------------
# 17. BASIC STATISTICS
# --------------------------------------------

print("\nPopulation Statistics for 2025:")
print(population_2025.describe())


# --------------------------------------------
# 18. FINAL OBSERVATION
# --------------------------------------------

print("\nObservation:")
print("""
The histogram shows the distribution of population across
countries and regions in 2025. The population is unevenly
distributed, with most countries having comparatively smaller
populations and a few countries having very large populations.

The bar chart highlights the 10 countries or regions with the
highest population in 2025.
""")
```
